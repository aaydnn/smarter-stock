```python
import pandas as pd

sheets = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name=None)  # 1–3 min

for name, s in sheets.items():
    s["source_sheet"] = name
    s["source_row"] = s.index + 2  # Excel row number (row 1 = header)

df = pd.concat(sheets.values(), ignore_index=True)
df["source_row_id"] = df["source_sheet"] + "-" + df["source_row"].astype(str)

print(df.shape, df["source_row_id"].is_unique)
```

```python
print(df.isna().sum(), "\n")
print("Cancellations:", df["Invoice"].astype(str).str.startswith("C").sum())
print("Qty <= 0:", (df["Quantity"] <= 0).sum())
print("Price <= 0:", (df["Price"] <= 0).sum())
print("Duplicates:", df.drop(columns=["source_sheet", "source_row", "source_row_id"]).duplicated().sum())
print("Date range:", df["InvoiceDate"].min(), "to", df["InvoiceDate"].max())
print("Products:", df["StockCode"].nunique())
print("Customers:", df["Customer ID"].nunique())
print("Countries:", df["Country"].nunique())
```

```python
for col in ["Invoice", "StockCode", "Description"]:
    df[col] = df[col].astype("string")

df.to_parquet("../data/raw/retail.parquet")
print("saved")
```
